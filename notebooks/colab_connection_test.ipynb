```python
from pathlib import Path

DRIVE_PROJECT = Path(
    "/content/drive/MyDrive/constellation-detection"
)

folders = [
    "data/raw",
    "data/processed",
    "data/annotations",
    "experiments",
    "models",
    "outputs",
]

for folder in folders:
    (DRIVE_PROJECT / folder).mkdir(
        parents=True,
        exist_ok=True
    )

print("Project storage initialized:")
for path in sorted(DRIVE_PROJECT.rglob("*")):
    if path.is_dir():
        print(path.relative_to(DRIVE_PROJECT))
```
